```python
# Notebook 01 - ESTRUTURAÇÃO E SINCRONIZAÇÃO DINÂMICA

import glob
import os
import pandas as pd

# 1. Definir o diretório onde estão os arquivos brutos de eye-tracking
diretorio_bruto = "../data/raw/"

# 2. Varredura automática por arquivos compatíveis (.csv e .xlsx)
arquivos_xlsx = glob.glob(os.path.join(diretorio_bruto, "*.xlsx"))
arquivos_csv = glob.glob(os.path.join(diretorio_bruto, "*.csv"))

lista_arquivos = arquivos_xlsx + arquivos_csv

print(f"Total de arquivos brutos encontrados: {len(lista_arquivos)}")

# 3. Construção do Manifesto Inicial
dados_manifesto = []

for caminho in lista_arquivos:
  nome_arquivo = os.path.basename(caminho)

  # Exemplo de extração de metadados a partir do nome do arquivo
  # (Assume-se um padrão como: ID_Condicao.csv ou similar.)
  partes = nome_arquivo.split(" ")
  sujeito_id = partes[0] if len(partes) > 0 else "Desconhecido"

  dados_manifesto.append(
      {"sujeito_id": sujeito_id, "nome_arquivo": nome_arquivo, "caminho_completo": caminho}
  )

df_manifesto = pd.DataFrame(dados_manifesto)

# 4. Visualizar as primeiras linhas do manifesto gerado
display(df_manifesto.head())

# 5. Salvar o manifesto para uso nos próximos notebooks
os.makedirs("../data/interim", exist_ok=True)
df_manifesto.to_csv("../data/interim/df_manifesto.csv", index=False)
print("Manifesto salvo com sucesso em '../data/interim/df_manifesto.csv'.")
```
